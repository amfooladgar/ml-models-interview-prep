## 🎯 Section 1: Overview

**Polynomial Regression** fits a polynomial relationship between features and target:

$$y = \theta_0 + \theta_1 x + \theta_2 x^2 + \theta_3 x^3 + \cdots + \theta_d x^d$$

### Key Insight
Polynomial regression is **still linear regression** — it's linear in the parameters θ! We just create new features (x², x³, ...) and apply linear regression to the expanded feature set.

### When to Use
- When the relationship between X and y is clearly nonlinear
- When you suspect a quadratic, cubic, or higher-order relationship
- As a stepping stone to understanding overfitting/underfitting

### Dangers
- **Overfitting** — high-degree polynomials fit training data perfectly but generalize poorly
- **Extrapolation** — polynomials behave wildly outside the training range

---
## 📐 Section 2: Math & Intuition

### Feature Transformation
Given a single feature x, we create polynomial features:

$$x \rightarrow [1, x, x^2, x^3, \ldots, x^d]$$

Then apply linear regression: $y = X_{poly} \cdot \theta$

### Bias-Variance Tradeoff

| | Low Degree (d=1) | High Degree (d=15) |
|---|---|---|
| **Bias** | High (underfitting) | Low |
| **Variance** | Low | High (overfitting) |
| **Training Error** | High | Very low |
| **Test Error** | High | High |

The **sweet spot** is where the sum of bias² + variance is minimized.

### How to Choose Degree
- **Cross-validation** — the gold standard
- **Learning curves** — plot train/test error vs degree
- **Domain knowledge** — physics often suggests quadratic/cubic

## ❓ Section 6: Interview Questions

### Q1: Is polynomial regression a linear or nonlinear model?
<details><summary>Click for answer</summary>

**It's a linear model** in terms of parameters! The features are nonlinear transforms of x (x², x³), but the model is still linear in θ. This means we can use the same optimization techniques (normal equation, gradient descent).

This is a very common interview trick question.
</details>

### Q2: How do you choose the degree of the polynomial?
<details><summary>Click for answer</summary>

1. **Cross-validation** (most reliable) — try different degrees, pick the one with best CV score
2. **Information criteria** — AIC or BIC penalize model complexity
3. **Domain knowledge** — physics/engineering often suggests the order
4. **Learning curves** — if train error ≪ test error, you're overfitting (reduce degree)
</details>

### Q3: What is the bias-variance tradeoff?
<details><summary>Click for answer</summary>

- **Bias**: Error from oversimplifying the model. High bias → underfitting.
- **Variance**: Error from model being too sensitive to training data. High variance → overfitting.
- **Tradeoff**: As model complexity increases, bias decreases but variance increases.
- **Goal**: Find the sweet spot where total error (bias² + variance + irreducible noise) is minimized.

$$\text{Total Error} = \text{Bias}^2 + \text{Variance} + \text{Irreducible Error}$$
</details>

### Q4: What happens if you use a very high degree polynomial?
<details><summary>Click for answer</summary>

- Model memorizes training data (training error → 0)
- Coefficients become very large
- Wild oscillations between data points (Runge's phenomenon)
- Terrible predictions on new data (high test error)
- Dangerous extrapolation behavior
- **Solution**: Use regularization (Ridge/Lasso) or reduce degree
</details>

### Q5: How does polynomial regression relate to feature engineering?
<details><summary>Click for answer</summary>

Polynomial regression is a specific case of **feature engineering** — creating new features from existing ones. Instead of using just x, we create x², x³, etc. This same idea extends to:
- Interaction terms (x₁ · x₂)
- Log/sqrt transformations
- Domain-specific features

The key insight: **a simple model with good features often beats a complex model with raw features.**
</details>
