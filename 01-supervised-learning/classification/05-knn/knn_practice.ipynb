## 🎯 Section 1: Overview & Math

### Distance Metrics
- **Euclidean Distance ($L_2$)**: $d(x, z) = \sqrt{\sum_{i=1}^{d} (x_i - z_i)^2}$
- **Manhattan Distance ($L_1$)**: $d(x, z) = \sum_{i=1}^{d} |x_i - z_i|$
- **Minkowski Distance ($L_p$)**: $d(x, z) = \left( \sum_{i=1}^{d} |x_i - z_i|^p \right)^{\frac{1}{p}}$

### Choice of K
- Small $k$ (e.g., $k=1$): High variance, sensitive to noise & outliers (overfitting).
- Large $k$: High bias, over-smoothed decision boundaries (underfitting).

## ❓ Section 4: Interview Questions

### Q1: Why is k-NN called a "Lazy Learner"?
<details><summary>Click for answer</summary>

k-NN does not have a training phase ($O(1)$ fit time). It simply stores the training instances in memory. All computation occurs at query/test time, where it computes distances from the target instance to every single stored training sample ($O(n \cdot d)$ prediction time).
</details>

### Q2: How does the Curse of Dimensionality affect k-NN?
<details><summary>Click for answer</summary>

In high-dimensional space, the volume of the space grows exponentially, causing data points to become extremely sparse. As a result, the distance between any two points approaches the same average distance, making the concept of "nearest neighbor" meaningless unless dimensionality reduction (PCA) is performed.
</details>
