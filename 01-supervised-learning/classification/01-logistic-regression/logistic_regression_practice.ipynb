## 🎯 Section 1: Overview

Despite its name, **Logistic Regression** is a **classification** algorithm used to predict binary outcomes (0 or 1, Yes or No, True or False).

### Why not Linear Regression for Classification?
1. Linear Regression outputs unbounded continuous values ($-\infty$ to $+\infty$), not probabilities $[0, 1]$.
2. Sensitive to outliers shifting the decision boundary.
3. Violates homoscedasticity assumptions.

## 📐 Section 2: Math & Intuition

### Sigmoid Function
$$\sigma(z) = \frac{1}{1 + e^{-z}}$$

Where $z = \theta^T x$. Maps any real number to $(0, 1)$.

### Hypothesis Function
$$h_\theta(x) = \sigma(\theta^T x) = \frac{1}{1 + e^{-\theta^T x}} = P(y=1|x; \theta)$$

### Cost Function: Binary Cross-Entropy (Log-Loss)
$$J(\theta) = -\frac{1}{m} \sum_{i=1}^{m} \left[ y^{(i)} \log(h_\theta(x^{(i)})) + (1 - y^{(i)}) \log(1 - h_\theta(x^{(i)})) \right]$$

### Gradient Update Rule
$$\theta := \theta - \frac{\alpha}{m} X^T(h_\theta(X) - y)$$

*(Notice the gradient form is identical to linear regression, but $h_\theta(x)$ contains the sigmoid function!)*

## ❓ Section 5: Interview Questions

### Q1: What is the cost function of Logistic Regression and why don't we use MSE?
<details><summary>Click for answer</summary>

If we use Mean Squared Error (MSE) with the Sigmoid function, the resulting cost function $J(\theta)$ is **non-convex** with many local minima due to the non-linear sigmoid transformation. Binary Cross-Entropy (Log-Loss) guarantees a **convex** cost function with a single global minimum that gradient descent is guaranteed to find.
</details>

### Q2: How do you extend Logistic Regression to Multi-Class Classification?
<details><summary>Click for answer</summary>

1. **One-vs-Rest (OvR / OvA)**: Train $K$ distinct binary logistic regression classifiers, each predicting class $k$ vs all other classes.
2. **Multinomial Logistic Regression (Softmax Regression)**: Generalize the Sigmoid function to the Softmax function over all $K$ classes.
</details>
