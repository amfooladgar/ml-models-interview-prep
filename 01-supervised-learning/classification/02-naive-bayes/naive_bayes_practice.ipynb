## 🎯 Section 1: Overview & Math

### Bayes' Theorem
$$P(y | x_1, x_2, \dots, x_n) = \frac{P(x_1, x_2, \dots, x_n | y) P(y)}{P(x_1, x_2, \dots, x_n)}$$

### The "Naive" Assumption
Assumes features $x_1, x_2, \dots, x_n$ are **mutually independent** given class $y$:
$$P(x_1, x_2, \dots, x_n | y) = \prod_{i=1}^{n} P(x_i | y)$$

### Classification Rule
$$\hat{y} = \arg\max_y P(y) \prod_{i=1}^{n} P(x_i | y)$$

*(In practice, we take log-likelihoods to avoid numerical underflow!)*
$$\hat{y} = \arg\max_y \left[ \log P(y) + \sum_{i=1}^{n} \log P(x_i | y) \right]$$

## ❓ Section 4: Interview Questions

### Q1: What is Laplace Smoothing and why is it necessary in Naive Bayes?
<details><summary>Click for answer</summary>

Laplace (additive) smoothing prevents zero probability estimation for unseen words/features during testing. If a feature $x_i$ never appears with class $y$ in training data, $P(x_i | y) = 0$, causing the entire product $\prod P(x_i | y)$ to become zero. Laplace smoothing adds $\alpha$ (typically $\alpha=1$) to the numerator and $\alpha \cdot |V|$ to the denominator.
</details>

### Q2: Why is Naive Bayes so fast and effective despite its unrealistic independence assumption?
<details><summary>Click for answer</summary>

1. **Computational Speed**: Training involves simple counting and mean/variance calculations ($O(n \cdot d)$ time). Prediction requires simple multiplications/additions.
2. **Effective Classification**: For classification, we only care about ranking $P(y|X)$, not accurate probability estimation. Even if the probabilities are biased due to feature dependence, the class with the maximum probability often remains correct.
</details>
