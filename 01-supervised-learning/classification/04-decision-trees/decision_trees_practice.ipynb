## 🎯 Section 1: Overview & Math

### Entropy
$$H(S) = -\sum_{i=1}^{K} p_i \log_2(p_i)$$

### Gini Impurity
$$G(S) = 1 - \sum_{i=1}^{K} p_i^2$$

### Information Gain
$$IG(S, A) = H(S) - \sum_{v \in \text{values}(A)} \frac{|S_v|}{|S|} H(S_v)$$

## ❓ Section 4: Interview Questions

### Q1: What is the difference between Gini Impurity and Entropy?
<details><summary>Click for answer</summary>

- **Gini Impurity** measures the probability of a randomly chosen element being incorrectly classified. Max value = $0.5$ (binary).
- **Entropy** measures the average amount of information or disorder. Max value = $1.0$ (binary).
- In practice, both yield almost identical trees. Gini is slightly faster to compute because it does not involve logarithmic calculations.
</details>

### Q2: Why are Decision Trees prone to overfitting and how can we prevent it?
<details><summary>Click for answer</summary>

Unconstrained decision trees can grow until every single leaf node is pure, effectively memorizing the training noise.
**Prevention**: Hyperparameter tuning (`max_depth`, `min_samples_split`, `min_samples_leaf`, `max_leaf_nodes`), Cost-Complexity Pruning ($\alpha$-pruning), or using Ensemble methods like Random Forests.
</details>
