## 🎯 Section 1: Overview

t-Distributed Stochastic Neighbor Embedding is a non-linear probabilistic visualization tool.

### Cost Function:
Kullback-Leibler divergence between high-dim $P$ and low-dim $Q$ distributions:
$$KL(P || Q) = \sum_{i} \sum_{j} p_{j|i} \log \frac{p_{j|i}}{q_{j|i}}$$ 
Using Student-t for low-dim avoids crowding.

## ❓ Section 4: Interview Questions


### Q1: What is the 'crowding problem' in dimensionality reduction, and how does t-SNE address it?
**Answer**: In high dimensions, there is more volume than in low dimensions. Moderate distances collapse and crowd together. t-SNE solves this by using Student-t distribution (heavier tails) in low dimensions.


### Q2: Can you use t-SNE to project out-of-sample test data?
**Answer**: No. t-SNE maps existing points dynamically. You cannot fit on train and transform on test; you must run t-SNE on the combined dataset.
