## ❓ Section 4: Interview Questions


### Q1: What is the maximum number of components LDA can extract?
**Answer**: Min(d, C-1) where d is number of features and C is number of classes. This is because class mean vectors span at most a C-1 dimensional subspace.


### Q2: How is LDA different from PCA?
**Answer**: PCA is unsupervised (ignores labels, maximizes overall variance). LDA is supervised (uses labels, maximizes class separability).
