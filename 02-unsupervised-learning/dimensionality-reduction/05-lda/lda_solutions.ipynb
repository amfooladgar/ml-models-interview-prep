## 🎯 Section 1: Overview

A supervised reduction method maximizing class separability by maximizing between-class scatter vs within-class scatter.

### Criterion:
$$J(w) = \frac{w^T S_B w}{w^T S_W w}$$


## 🔧 Section 2: Implementation from Scratch


```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

np.random.seed(42)
plt.style.use('seaborn-v0_8-whitegrid')
sns.set_palette('husl')

print('Setup complete! ✅')
```

```python
class LDAFromScratch:
    def __init__(self, n_components=1):
        self.n_components = n_components
        
    def fit_transform(self, X, y):
        n_features = X.shape[1]
        class_labels = np.unique(y)
        
        mean_overall = np.mean(X, axis=0)
        SW = np.zeros((n_features, n_features))
        SB = np.zeros((n_features, n_features))
        
        for c in class_labels:
            X_c = X[y == c]
            mean_c = np.mean(X_c, axis=0)
            SW += np.dot((X_c - mean_c).T, (X_c - mean_c))
            
            n_c = X_c.shape[0]
            mean_diff = (mean_c - mean_overall).reshape(-1, 1)
            SB += n_c * np.dot(mean_diff, mean_diff.T)
            
        # Generalized eigenvalue problem: inv(SW) @ SB
        A = np.dot(np.linalg.pinv(SW), SB)
        eigenvalues, eigenvectors = np.linalg.eigh(A)
        
        sorted_indices = np.argsort(eigenvalues)[::-1]
        self.w = eigenvectors[:, sorted_indices[:self.n_components]]
        return np.dot(X, self.w)
```

```python
# Verify that the implementation runs and outputs correctly
X = np.random.rand(20, 3)
y = np.array([0]*10 + [1]*10)
model = LDAFromScratch(n_components=1)
proj = model.fit_transform(X, y)
print('LDA projected shape:', proj.shape)
```

## ❓ Section 4: Interview Questions


### Q1: What is the maximum number of components LDA can extract?
**Answer**: Min(d, C-1) where d is number of features and C is number of classes. This is because class mean vectors span at most a C-1 dimensional subspace.


### Q2: How is LDA different from PCA?
**Answer**: PCA is unsupervised (ignores labels, maximizes overall variance). LDA is supervised (uses labels, maximizes class separability).
