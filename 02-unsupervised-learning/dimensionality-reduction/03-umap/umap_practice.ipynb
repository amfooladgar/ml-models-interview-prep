### Q1: What are UMAP's two main hyperparameters?
**Answer**: `n_neighbors` (size of local neighborhood, balancing local vs global focus) and `min_dist` (minimum distance between points in low dimensions, controlling packing density).


### Q2: How does UMAP compare to t-SNE?
