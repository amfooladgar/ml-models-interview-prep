## 🎯 Section 1: Overview

UMAP is a manifold learning technique based on Riemannian geometry that preserves local and global structures.

### Core Philosophy:
UMAP models the manifold using fuzzy simplicial sets and projects it by minimizing fuzzy set cross-entropy.

## ❓ Section 4: Interview Questions


### Q1: What are UMAP's two main hyperparameters?
**Answer**: `n_neighbors` (size of local neighborhood, balancing local vs global focus) and `min_dist` (minimum distance between points in low dimensions, controlling packing density).


### Q2: How does UMAP compare to t-SNE?
**Answer**: UMAP preserves global structure better, is significantly faster on large datasets, and allows projecting new out-of-sample data points.
