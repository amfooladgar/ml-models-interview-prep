## 🎯 Section 1: Overview

A linear dimensionality reduction technique that finds orthogonal directions of maximum variance.

### Objective:
Maximize variance of projected data. Covariance matrix: $\Sigma = \frac{1}{m} X^T X$. Solve: $\Sigma v = \lambda v$.


## 🔧 Section 2: Implementation from Scratch


```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

np.random.seed(42)
plt.style.use('seaborn-v0_8-whitegrid')
sns.set_palette('husl')

print('Setup complete! ✅')
```

```python
class PCAFromScratch:
    def __init__(self, n_components=2):
        self.n_components = n_components
        
    def fit_transform(self, X):
        # Center the data
        self.mean_ = np.mean(X, axis=0)
        X_centered = X - self.mean_
        
        # Covariance matrix
        cov = np.cov(X_centered, rowvar=False)
        
        # Eigenvalues and eigenvectors
        eigenvalues, eigenvectors = np.linalg.eigh(cov)
        
        # Sort in descending order
        sorted_indices = np.argsort(eigenvalues)[::-1]
        self.components_ = eigenvectors[:, sorted_indices[:self.n_components]].T
        
        # Project data
        return np.dot(X_centered, self.components_.T)
```

```python
# Verify that the implementation runs and outputs correctly
X = np.random.rand(50, 4)
model = PCAFromScratch(n_components=2)
proj = model.fit_transform(X)
print('Projected shape:', proj.shape)
```

## ❓ Section 4: Interview Questions


### Q1: Why should you standardize features before running PCA?
**Answer**: PCA is sensitive to feature variances. If one feature is measured in thousands and another in decimals, the eigenvalue calculation will be dominated by the larger-scale feature.


### Q2: What is the explained variance ratio?
**Answer**: The fraction of the dataset's total variance that lies along each principal component axis, calculated as lambda_i / sum(lambda).
