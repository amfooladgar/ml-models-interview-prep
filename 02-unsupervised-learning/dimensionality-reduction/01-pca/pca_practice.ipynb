## 🎯 Section 1: Overview

A linear dimensionality reduction technique that finds orthogonal directions of maximum variance.

### Objective:
Maximize variance of projected data. Covariance matrix: $\Sigma = \frac{1}{m} X^T X$. Solve: $\Sigma v = \lambda v$.

## ❓ Section 4: Interview Questions


### Q1: Why should you standardize features before running PCA?
**Answer**: PCA is sensitive to feature variances. If one feature is measured in thousands and another in decimals, the eigenvalue calculation will be dominated by the larger-scale feature.


### Q2: What is the explained variance ratio?
**Answer**: The fraction of the dataset's total variance that lies along each principal component axis, calculated as lambda_i / sum(lambda).
