## 🎯 Section 1: Overview

Decomposes any real matrix into orthogonal vectors representing singular components.

### Decomposition:
$$A = U \Sigma V^T$$


## 🔧 Section 2: Implementation from Scratch


```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

np.random.seed(42)
plt.style.use('seaborn-v0_8-whitegrid')
sns.set_palette('husl')

print('Setup complete! ✅')
```

```python
class SVDFromScratch:
    def __init__(self, n_components=2):
        self.n_components = n_components
        
    def fit_transform(self, A):
        # ATA has shape (n, n)
        ATA = np.dot(A.T, A)
        eigenvalues, V = np.linalg.eigh(ATA)
        
        # Sort in descending order
        sorted_indices = np.argsort(eigenvalues)[::-1]
        V_truncated = V[:, sorted_indices[:self.n_components]]
        
        # Project to lower dimension
        return np.dot(A, V_truncated)
```

```python
# Verify that the implementation runs and outputs correctly
X = np.random.rand(10, 5)
model = SVDFromScratch(n_components=2)
proj = model.fit_transform(X)
print('SVD projected shape:', proj.shape)
```

## 📦 Section 3: Library Implementation


```python
from sklearn.decomposition import TruncatedSVD
svd = TruncatedSVD(n_components=5, random_state=42)
X_reduced = svd.fit_transform(X)
```

## ❓ Section 4: Interview Questions


### Q1: What is the difference between SVD and PCA?
**Answer**: PCA is equivalent to performing SVD on the centered covariance matrix. SVD can be run directly on the raw data matrix, which is highly useful for sparse matrices.


### Q2: What is Truncated SVD used for in NLP?
**Answer**: Latent Semantic Analysis (LSA). It reduces term-document matrices to discover semantic topics.
