## 🎯 Section 1: Overview

Eclat uses vertical database layout (TID-lists) to perform set intersection for frequent itemsets.

### Vertical format:
Item -> Set of Transactions [TID1, TID2, ...]. Support computed via transaction set intersections.


## 🔧 Section 2: Implementation from Scratch


```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

np.random.seed(42)
plt.style.use('seaborn-v0_8-whitegrid')
sns.set_palette('husl')

print('Setup complete! ✅')
```

```python
class EclatFromScratch:
    def __init__(self, min_support=0.1):
        self.min_support = min_support
        
    def fit(self, transactions):
        n_trans = len(transactions)
        # Vertical representation
        vertical_data = {}
        for tid, t in enumerate(transactions):
            for item in t:
                if item not in vertical_data:
                    vertical_data[item] = set()
                vertical_data[item].add(tid)
                
        # Find frequent itemsets recursively
        frequent_itemsets = {}
        def eclat(prefix, items):
            while items:
                item, tids = items.pop(0)
                support = len(tids) / n_trans
                if support >= self.min_support:
                    new_prefix = prefix + [item]
                    frequent_itemsets[frozenset(new_prefix)] = support
                    
                    suffix = []
                    for other_item, other_tids in items:
                        intersection = tids.intersection(other_tids)
                        if len(intersection) / n_trans >= self.min_support:
                            suffix.append((other_item, intersection))
                    eclat(new_prefix, suffix)
                    
        items_list = [(item, tids) for item, tids in vertical_data.items()]
        eclat([], items_list)
        self.frequent_itemsets_ = frequent_itemsets
        return self
```

```python
# Verify that the implementation runs and outputs correctly
transactions = [[1, 2, 5], [2, 4], [2, 3], [1, 2, 4], [1, 3], [2, 3], [1, 3], [1, 2, 3, 5], [1, 2, 3]]
model = EclatFromScratch(min_support=0.2)
model.fit(transactions)
print('Eclat frequent count:', len(model.frequent_itemsets_))
```

## ❓ Section 4: Interview Questions


### Q1: How does Eclat differ from Apriori?
**Answer**: Apriori uses horizontal transaction scanning and candidate generation. Eclat uses a vertical database format (TID-lists) and calculates support using set intersections, avoiding full scans.


### Q2: What is the advantage of Eclat's vertical intersection?
**Answer**: It is much faster than scanning the horizontal layout, especially if TID lists fit in memory.
