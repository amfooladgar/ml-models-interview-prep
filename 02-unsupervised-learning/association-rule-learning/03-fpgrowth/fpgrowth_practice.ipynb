## ❓ Section 4: Interview Questions


### Q1: Why is FP-Growth faster than Apriori?
**Answer**: FP-Growth does not generate any candidate itemsets. It scans the database only twice to build the FP-Tree and then recursively mines it, making it extremely memory-efficient and fast.


### Q2: What are the components of an FP-Tree?
**Answer**: A root node, children nodes representing prefix item paths, and a Header Table containing pointers to all nodes of each item for fast traversal.
