## 🎯 Section 1: Overview

Iteratively generates candidate itemsets and prunes them using support threshold.

### Apriori Principle:
Any subset of a frequent itemset must also be frequent. If an itemset is infrequent, all its supersets are infrequent.


## 🔧 Section 2: Implementation from Scratch


```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

np.random.seed(42)
plt.style.use('seaborn-v0_8-whitegrid')
sns.set_palette('husl')

print('Setup complete! ✅')
```

```python
from collections import defaultdict

class AprioriFromScratch:
    def __init__(self, min_support=0.1):
        self.min_support = min_support
        
    def fit(self, transactions):
        n_trans = len(transactions)
        # Frequent 1-itemsets
        item_counts = defaultdict(int)
        for t in transactions:
            for item in t:
                item_counts[item] += 1
        
        frequent = {frozenset([item]): count/n_trans for item, count in item_counts.items() if count/n_trans >= self.min_support}
        all_frequent = dict(frequent)
        
        k = 2
        while frequent:
            # Generate candidate (k)-itemsets
            items = list(frequent.keys())
            candidates = set()
            for i in range(len(items)):
                for j in range(i+1, len(items)):
                    union = items[i].union(items[j])
                    if len(union) == k:
                        candidates.add(union)
            
            # Prune and compute support
            cand_counts = defaultdict(int)
            for t in transactions:
                for cand in candidates:
                    if cand.issubset(t):
                        cand_counts[cand] += 1
            
            frequent = {cand: count/n_trans for cand, count in cand_counts.items() if count/n_trans >= self.min_support}
            all_frequent.update(frequent)
            k += 1
            
        self.frequent_itemsets_ = all_frequent
        return self
```

```python
# Verify that the implementation runs and outputs correctly
transactions = [[1, 2, 5], [2, 4], [2, 3], [1, 2, 4], [1, 3], [2, 3], [1, 3], [1, 2, 3, 5], [1, 2, 3]]
model = AprioriFromScratch(min_support=0.2)
model.fit(transactions)
print('Frequent itemset count:', len(model.frequent_itemsets_))
```

### Q1: Define Support, Confidence, and Lift.
**Answer**: - Support: proportion of transactions containing the itemset. - Confidence: P(B|A) = support(A U B) / support(A). - Lift: support(A U B) / (support(A) * support(B)).
