## 🎯 Section 1: Overview

DBSCAN groups points based on spatial density and flags sparse points as noise/outliers.

### Density Concepts:
- **Core point**: $\geq$ `min_samples` within $\epsilon$.
- **Border point**: $<`min_samples` within $\epsilon$, but in neighborhood of a Core point.
- **Noise**: All other points.


## 🔧 Section 2: Implementation from Scratch


```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

np.random.seed(42)
plt.style.use('seaborn-v0_8-whitegrid')
sns.set_palette('husl')

print('Setup complete! ✅')
```

```python
class DBSCANFromScratch:
    def __init__(self, eps=0.5, min_samples=5):
        self.eps = eps
        self.min_samples = min_samples
        
    def fit(self, X):
        n_samples = X.shape[0]
        self.labels_ = np.full(n_samples, -1)  # Initialize as noise (-1)
        visited = np.zeros(n_samples, dtype=bool)
        
        def get_neighbors(i):
            return np.where(np.linalg.norm(X - X[i], axis=1) < self.eps)[0]
            
        cluster_id = 0
        for i in range(n_samples):
            if visited[i]:
                continue
            visited[i] = True
            neighbors = get_neighbors(i)
            
            if len(neighbors) < self.min_samples:
                self.labels_[i] = -1
            else:
                self.labels_[i] = cluster_id
                # Expand cluster
                queue = list(neighbors)
                for item in queue:
                    if not visited[item]:
                        visited[item] = True
                        item_neighbors = get_neighbors(item)
                        if len(item_neighbors) >= self.min_samples:
                            queue.extend(item_neighbors)
                    if self.labels_[item] == -1:
                        self.labels_[item] = cluster_id
                cluster_id += 1
        return self
```

```python
# Verify that the implementation runs and outputs correctly
X = np.random.rand(50, 2)
model = DBSCANFromScratch(eps=0.2, min_samples=3)
model.fit(X)
print('Noise count:', np.sum(model.labels_ == -1))
```

## 📦 Section 3: Library Implementation


```python
from sklearn.cluster import DBSCAN
dbscan = DBSCAN(eps=0.5, min_samples=5)
labels = dbscan.fit_predict(X)
```

## ❓ Section 4: Interview Questions


### Q1: What are the advantages of DBSCAN over K-Means?
**Answer**: Doesn't require specifying cluster count beforehand, handles arbitrary shapes, and explicitly identifies outliers/noise.


### Q2: What is the time complexity of DBSCAN?
**Answer**: O(N^2) in worst case, but can be reduced to O(N log N) using spatial index structures like R-Trees/KD-Trees for neighborhood queries.
