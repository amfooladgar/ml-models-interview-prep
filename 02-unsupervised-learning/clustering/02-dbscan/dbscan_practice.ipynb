## 🎯 Section 1: Overview

DBSCAN groups points based on spatial density and flags sparse points as noise/outliers.

### Density Concepts:
- **Core point**: $\geq$ `min_samples` within $\epsilon$.
- **Border point**: $<`min_samples` within $\epsilon$, but in neighborhood of a Core point.
- **Noise**: All other points.

## ❓ Section 4: Interview Questions


### Q1: What are the advantages of DBSCAN over K-Means?
**Answer**: Doesn't require specifying cluster count beforehand, handles arbitrary shapes, and explicitly identifies outliers/noise.


### Q2: What is the time complexity of DBSCAN?
**Answer**: O(N^2) in worst case, but can be reduced to O(N log N) using spatial index structures like R-Trees/KD-Trees for neighborhood queries.
