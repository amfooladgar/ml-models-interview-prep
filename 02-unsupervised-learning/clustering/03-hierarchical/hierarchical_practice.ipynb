## 🎯 Section 1: Overview

Builds a tree of clusters bottom-up (Agglomerative) or top-down (Divisive) based on linkage criteria.

### Linkage Types:
- **Single Linkage**: Min distance between elements in two clusters.
- **Complete Linkage**: Max distance.
- **Average Linkage**: Mean distance.
- **Ward**: Minimizes total within-cluster variance.

## ❓ Section 4: Interview Questions


### Q1: What is a dendrogram?
**Answer**: A tree diagram illustrating the arrangement of clusters produced by hierarchical clustering, where the height of a branch represents distance/dissimilarity between merged clusters.


### Q2: What linkage minimizes intra-cluster variance?
**Answer**: Ward's linkage minimizes the sum of squared distances within clusters.
