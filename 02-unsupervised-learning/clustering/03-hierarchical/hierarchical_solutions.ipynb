## 🎯 Section 1: Overview

Builds a tree of clusters bottom-up (Agglomerative) or top-down (Divisive) based on linkage criteria.

### Linkage Types:
- **Single Linkage**: Min distance between elements in two clusters.
- **Complete Linkage**: Max distance.
- **Average Linkage**: Mean distance.
- **Ward**: Minimizes total within-cluster variance.


## 🔧 Section 2: Implementation from Scratch


```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

np.random.seed(42)
plt.style.use('seaborn-v0_8-whitegrid')
sns.set_palette('husl')

print('Setup complete! ✅')
```

```python
class AgglomerativeFromScratch:
    def __init__(self, n_clusters=2):
        self.n_clusters = n_clusters
        
    def fit(self, X):
        n_samples = X.shape[0]
        # Initialize clusters: each point is its own cluster
        clusters = {i: [i] for i in range(n_samples)}
        
        while len(clusters) > self.n_clusters:
            # Find closest clusters
            min_dist = float('inf')
            to_merge = (None, None)
            
            # Simple single linkage for demonstration
            for c1_id, c1 in clusters.items():
                for c2_id, c2 in clusters.items():
                    if c1_id >= c2_id:
                        continue
                    for idx1 in c1:
                        for idx2 in c2:
                            dist = np.linalg.norm(X[idx1] - X[idx2])
                            if dist < min_dist:
                                min_dist = dist
                                to_merge = (c1_id, c2_id)
            
            c1, c2 = to_merge
            clusters[c1].extend(clusters[c2])
            del clusters[c2]
            
        self.labels_ = np.zeros(n_samples)
        for cluster_id, (c_key, indices) in enumerate(clusters.items()):
            for idx in indices:
                self.labels_[idx] = cluster_id
        return self
```

```python
# Verify that the implementation runs and outputs correctly
X = np.random.rand(20, 2)
model = AgglomerativeFromScratch(n_clusters=2)
model.fit(X)
print('Cluster assignments:', model.labels_)
```

## 📦 Section 3: Library Implementation


```python
from sklearn.cluster import AgglomerativeClustering
agg = AgglomerativeClustering(n_clusters=3, linkage='ward')
labels = agg.fit_predict(X)
```

## ❓ Section 4: Interview Questions


### Q1: What is a dendrogram?
**Answer**: A tree diagram illustrating the arrangement of clusters produced by hierarchical clustering, where the height of a branch represents distance/dissimilarity between merged clusters.


### Q2: What linkage minimizes intra-cluster variance?
**Answer**: Ward's linkage minimizes the sum of squared distances within clusters.
