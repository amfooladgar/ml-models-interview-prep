## 🎯 Section 1: Overview

K-Means partitioning algorithm groups points into K clusters by minimizing within-cluster variance.

### Cost Function (Inertia):
$$J = \sum_{j=1}^K \sum_{i \in S_j} \|x^{(i)} - \mu_j\|^2$$


## 🔧 Section 2: Implementation from Scratch


```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

np.random.seed(42)
plt.style.use('seaborn-v0_8-whitegrid')
sns.set_palette('husl')

print('Setup complete! ✅')
```

```python
class KMeansFromScratch:
    def __init__(self, k=3, max_iters=100):
        self.k = k
        self.max_iters = max_iters
        
    def fit(self, X):
        n_samples, n_features = X.shape
        # Initialize randomly
        random_idx = np.random.choice(n_samples, self.k, replace=False)
        self.centroids = X[random_idx]
        
        for _ in range(self.max_iters):
            # Assign clusters
            distances = np.linalg.norm(X[:, np.newaxis] - self.centroids, axis=2)
            labels = np.argmin(distances, axis=1)
            
            # Update centroids
            new_centroids = np.array([X[labels == j].mean(axis=0) if len(X[labels == j]) > 0 else self.centroids[j] for j in range(self.k)])
            if np.allclose(self.centroids, new_centroids):
                break
            self.centroids = new_centroids
        self.labels_ = labels
        return self
```

```python
# Verify that the implementation runs and outputs correctly
X = np.random.rand(100, 2)
model = KMeansFromScratch(k=3)
model.fit(X)
print('Centroids:\n', model.centroids)
```

## 📦 Section 3: Library Implementation


```python
from sklearn.cluster import KMeans
kmeans = KMeans(n_clusters=3, random_state=42)
labels = kmeans.fit_predict(X)
```

## ❓ Section 4: Interview Questions


### Q1: How do you choose the value of K in K-Means?
**Answer**: Using the Elbow Method (plotting inertia vs K and looking for the 'elbow') or Silhouette Analysis (evaluating cluster separation).


### Q2: Is K-Means sensitive to outliers?
**Answer**: Yes, because centroids are updated as the mean of points, and outliers pull the mean significantly, distorting the cluster shape.
