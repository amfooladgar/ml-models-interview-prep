## 🎯 Section 1: Overview

LightGBM grows trees leaf-wise and uses Gradient-based One-Side Sampling (GOSS) & Exclusive Feature Bundling (EFB) for massive speedups.

### GOSS & EFB:
- **GOSS**: Retains top instances with large gradients, samples small gradient instances.
- **EFB**: Combines sparse mutually exclusive features.


## 🔧 Section 2: Implementation from Scratch


```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

np.random.seed(42)
plt.style.use('seaborn-v0_8-whitegrid')
sns.set_palette('husl')

print('Setup complete! ✅')
```

```python
import lightgbm as lgb

X_train, y_train = np.random.rand(200, 10), np.random.randint(0, 2, 200)

lgbm = lgb.LGBMClassifier(
    n_estimators=100,
    learning_rate=0.05,
    num_leaves=31,
    subsample=0.8,
    colsample_bytree=0.8,
    random_state=42
)
lgbm.fit(X_train, y_train)
print("LightGBM model fitted! Score:", lgbm.score(X_train, y_train))
```

## ❓ Section 4: Interview Questions


### Q1: What is the difference between level-wise and leaf-wise tree growth?
**Answer**: Level-wise (XGBoost/GBM default) splits all nodes at the current depth uniformly. Leaf-wise (LightGBM) splits the node with max loss reduction, achieving lower loss faster but risking overfitting on small datasets.


### Q2: How does LightGBM handle categorical features?
**Answer**: It handles them directly without one-hot encoding by sorting categories according to target statistics and finding split thresholds in O(K log K).
