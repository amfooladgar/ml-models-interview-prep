## ❓ Section 4: Interview Questions


### Q1: What is target leakage in categorical encoding, and how does CatBoost solve it?
**Answer**: Standard target encoding uses the target label of the sample to compute feature statistics, causing data leakage. CatBoost uses Ordered Target Statistics based on random data permutations to compute stats using only prior samples.


### Q2: What are Oblivious / Symmetric Trees in CatBoost?
**Answer**: Trees where every node at depth $d$ uses the exact same feature and split value. This acts as a regularizer and enables hyper-fast CPU evaluation via SIMD instructions.
