## 🎯 Section 1: Overview

GBM builds trees sequentially to fit pseudo-residuals (negative gradients of loss function) of previous predictions.

### Pseudo-residuals:
$$r_{i,t} = -\left[ \frac{\partial L(y_i, f(x_i))}{\partial f(x_i)} \right]_{f_{t-1}(x_i)}$$


## 🔧 Section 2: Implementation from Scratch


```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

np.random.seed(42)
plt.style.use('seaborn-v0_8-whitegrid')
sns.set_palette('husl')

print('Setup complete! ✅')
```

```python
from sklearn.tree import DecisionTreeRegressor

class GradientBoostingRegressorFromScratch:
    def __init__(self, n_estimators=100, learning_rate=0.1, max_depth=3):
        self.n_estimators = n_estimators
        self.learning_rate = learning_rate
        self.max_depth = max_depth
        self.trees = []
        
    def fit(self, X, y):
        self.init_value = np.mean(y)
        f_pred = np.full(y.shape, self.init_value)
        
        for _ in range(self.n_estimators):
            # Compute negative gradient (residuals for MSE loss)
            residuals = y - f_pred
            
            tree = DecisionTreeRegressor(max_depth=self.max_depth, random_state=42)
            tree.fit(X, residuals)
            
            update = tree.predict(X)
            f_pred += self.learning_rate * update
            self.trees.append(tree)
        return self

    def predict(self, X):
        preds = np.full(X.shape[0], self.init_value)
        for tree in self.trees:
            preds += self.learning_rate * tree.predict(X)
        return preds
```

```python
X = np.random.rand(100, 4)
y = np.random.rand(100)
gbm = GradientBoostingRegressorFromScratch(n_estimators=20)
gbm.fit(X, y)
print('GBM predictions shape:', gbm.predict(X).shape)
```

## ❓ Section 4: Interview Questions


### Q1: Why is learning rate (shrinkage) used in Gradient Boosting?
**Answer**: Learning rate scales down the contribution of each individual tree, requiring more trees to fit the training data. This acts as a regularizer to prevent overfitting.


### Q2: How does GBM handle classification vs regression?
**Answer**: It uses different loss functions. Regression typically uses MSE or MAE; classification uses Binary Cross-Entropy (Log-Loss) or Multi-class Log-Loss.
